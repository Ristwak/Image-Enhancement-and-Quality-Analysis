```python
from google.colab import drive
drive.mount('/content/drive')

import cv2
import numpy as np
import os

# ====== INPUT / OUTPUT PATHS ======
input_folder = "/content/drive/MyDrive/Colab Notebooks/data/test_images"          # folder with original images
output_folder = "/content/drive/MyDrive/Colab Notebooks/data/low_quality_images"  # folder to save degraded images

os.makedirs(output_folder, exist_ok=True)

# ====== FUNCTIONS ======

def add_gaussian_noise(image):
    mean = 0
    std = 25
    noise = np.random.normal(mean, std, image.shape).astype(np.float32)
    noisy = image.astype(np.float32) + noise
    noisy = np.clip(noisy, 0, 255)
    return noisy.astype(np.uint8)

def apply_blur(image):
    return cv2.GaussianBlur(image, (9, 9), 0)

def reduce_resolution(image):
    h, w = image.shape[:2]
    small = cv2.resize(image, (w//4, h//4))   # downscale
    return cv2.resize(small, (w, h))          # upscale back

# ====== PROCESS IMAGES ======

for filename in os.listdir(input_folder):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        img_path = os.path.join(input_folder, filename)
        image = cv2.imread(img_path)

        if image is None:
            continue

        # Apply degradations
        noisy = add_gaussian_noise(image)
        blurred = apply_blur(noisy)
        low_quality = reduce_resolution(blurred)

        # Save image
        save_path = os.path.join(output_folder, filename)
        cv2.imwrite(save_path, low_quality)

        print(f"Processed: {filename}")

print("All images processed and saved!")
```
